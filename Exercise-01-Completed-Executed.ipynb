## **Dirty Dataset**

```python
# load the dirty dataset csv file into a pandas dataframe
fetal_dirty_df = pd.read_csv('fetal_health_dataset_dirty.csv')
```

### Problem 1.2

#### (a)

```python
# determine whether any NaN value is present
print(fetal_dirty_df.isna().any().any())
```

#### (b)

```python
# list variables that contain at least one missing value
missing_by_variable = fetal_dirty_df.isna().sum()
missing_variables = missing_by_variable[missing_by_variable > 0]
print(missing_variables)
```

#### (c)

```python
# count samples (rows) containing at least one missing value
missing_sample_count_before = fetal_dirty_df.isna().any(axis=1).sum()
print('Samples with at least one missing value:', missing_sample_count_before)
```

### Problem 1.3

#### (a)

```python
# Work on a copy so the initially loaded dirty dataframe is preserved.
fetal_valid_df = fetal_dirty_df.copy()

# Fetal-heart-rate baseline must be positive.
fetal_valid_df.loc[fetal_valid_df['baseline value'] <= 0, 'baseline value'] = np.nan

# Event rates/count-like measurements cannot be negative.
nonnegative_columns = [
    'accelerations', 'fetal_movement', 'uterine_contractions',
    'light_decelerations', 'severe_decelerations',
    'prolongued_decelerations', 'mean_value_of_short_term_variability',
    'mean_value_of_long_term_variability', 'histogram_width',
    'histogram_number_of_peaks', 'histogram_number_of_zeroes',
    'histogram_variance'
]
for column in nonnegative_columns:
    fetal_valid_df.loc[fetal_valid_df[column] < 0, column] = np.nan

# Percentage variables must lie between 0 and 100.
percentage_columns = [
    'abnormal_short_term_variability',
    'percentage_of_time_with_abnormal_long_term_variability'
]
for column in percentage_columns:
    invalid = ~fetal_valid_df[column].between(0, 100) & fetal_valid_df[column].notna()
    fetal_valid_df.loc[invalid, column] = np.nan

# Histogram FHR values should be positive.
histogram_value_columns = [
    'histogram_min', 'histogram_max', 'histogram_mode',
    'histogram_mean', 'histogram_median'
]
for column in histogram_value_columns:
    fetal_valid_df.loc[fetal_valid_df[column] <= 0, column] = np.nan

# Categorical codes must belong to their documented sets.
fetal_valid_df.loc[~fetal_valid_df['histogram_tendency'].isin([-1, 0, 1]), 'histogram_tendency'] = np.nan
fetal_valid_df.loc[~fetal_valid_df['fetal_health'].isin([1, 2, 3]), 'fetal_health'] = np.nan

# Show requested rows after validation.
fetal_valid_df.loc[91:100]
```

The validation rules follow the physical meaning of the variables. A fetal-heart-rate baseline and histogram heart-rate values cannot be zero or negative; event rates/counts and variances cannot be negative; percentage variables must be within $[0,100]$; and the two categorical variables must use only their documented category codes. Applying these rules reveals negative values in `baseline value`, which are physically impossible and are therefore replaced with `NaN`.


#### (b)

```python
# compare missingness before and after logical validation
missing_entries_before = fetal_dirty_df.isna().sum().sum()
missing_samples_before = fetal_dirty_df.isna().any(axis=1).sum()

missing_entries_after = fetal_valid_df.isna().sum().sum()
missing_samples_after = fetal_valid_df.isna().any(axis=1).sum()

print('Before validation - missing entries:', missing_entries_before)
print('Before validation - samples with missing entries:', missing_samples_before)
print('After validation  - missing entries:', missing_entries_after)
print('After validation  - samples with missing entries:', missing_samples_after)
```

Before logical validation, the dirty dataframe has **86 missing entries in 86 samples**. After invalid values are converted to `NaN`, it has **97 missing entries in 89 samples**. The number of missing entries increases by 11 because 11 negative `baseline value` measurements are invalid. The number of affected samples increases by only 3 because 8 of those 11 rows already contained another missing value.


### Problem 1.4

```python
# Standardize each non-categorical independent variable using sample std (ddof=1).
fetal_processed_df = fetal_valid_df.copy()

fetal_categorical = ['histogram_tendency', 'fetal_health']
fetal_numeric_features = [
    c for c in fetal_processed_df.columns
    if c not in fetal_categorical
]

for column in fetal_numeric_features:
    mean = fetal_processed_df[column].mean(skipna=True)
    std = fetal_processed_df[column].std(skipna=True, ddof=1)
    fetal_processed_df[column] = (fetal_processed_df[column] - mean) / std

fetal_processed_df.head(20)
```

### Problem 1.5

```python
# Replace NaN values with the median from the same fetal-health category.
for column in fetal_processed_df.columns:
    if column != 'fetal_health':
        group_median = fetal_processed_df.groupby('fetal_health')[column].transform('median')
        fetal_processed_df[column] = fetal_processed_df[column].fillna(group_median)

# Confirm that all replaceable missing values are gone.
print(fetal_processed_df.isna().sum())
```

### Problem 1.6

```python
# One-hot encode fetal_health and save the final processed dataset.
# The target has no missing values here, so convert its labels back to integers first.
fetal_processed_df['fetal_health'] = fetal_processed_df['fetal_health'].astype(int)

fetal_final_df = pd.get_dummies(
    fetal_processed_df,
    columns=['fetal_health'],
    prefix='fetal_health',
    dtype=int
)

fetal_final_df.to_csv('fetal_health_dataset_processed.csv', index=False)
print(fetal_final_df.head())
print('Saved fetal_health_dataset_processed.csv')
```

## **Preprocessing**

### Problem 2.2

```python
# Validate logical consistency using the documented meaning of each variable.
heart_valid_df = heart_df.copy()

# Positive/physically nonnegative numerical quantities.
heart_valid_df.loc[heart_valid_df['age'] <= 0, 'age'] = np.nan
heart_valid_df.loc[heart_valid_df['creatinine_phosphokinase'] < 0, 'creatinine_phosphokinase'] = np.nan
heart_valid_df.loc[~heart_valid_df['ejection_fraction'].between(0, 100), 'ejection_fraction'] = np.nan
heart_valid_df.loc[heart_valid_df['platelets'] < 0, 'platelets'] = np.nan
heart_valid_df.loc[heart_valid_df['serum_creatinine'] < 0, 'serum_creatinine'] = np.nan
heart_valid_df.loc[heart_valid_df['serum_sodium'] < 0, 'serum_sodium'] = np.nan
heart_valid_df.loc[heart_valid_df['time'] < 0, 'time'] = np.nan

# Binary variables must contain only 0 or 1.
binary_columns = [
    'anaemia', 'diabetes', 'high_blood_pressure',
    'sex', 'smoking', 'DEATH_EVENT'
]
for column in binary_columns:
    invalid = ~heart_valid_df[column].isin([0, 1]) & heart_valid_df[column].notna()
    heart_valid_df.loc[invalid, column] = np.nan

print('Total missing entries after validation:', heart_valid_df.isna().sum().sum())
```

The consistency checks use the documented meaning of each measurement. Age must be positive; laboratory/count measurements and follow-up time cannot be negative; ejection fraction is a percentage and must lie within $[0,100]$; and the six binary variables must be coded as 0 or 1. No values in the provided heart-failure dataset violate these rules, so this validation step does not introduce any missing values.


### Problem 2.3

```python
# Standardize non-categorical independent variables only.
heart_processed_df = heart_valid_df.copy()

heart_categorical = [
    'anaemia', 'diabetes', 'high_blood_pressure',
    'sex', 'smoking', 'DEATH_EVENT'
]
heart_numeric_features = [
    c for c in heart_processed_df.columns
    if c not in heart_categorical
]

for column in heart_numeric_features:
    mean = heart_processed_df[column].mean(skipna=True)
    std = heart_processed_df[column].std(skipna=True, ddof=1)
    heart_processed_df[column] = (heart_processed_df[column] - mean) / std

heart_processed_df.head(20)
```

### Problem 2.4

```python
# One-hot encode DEATH_EVENT while preserving NaN targets as NaN in both columns.
death_0 = pd.Series(np.nan, index=heart_processed_df.index, dtype=float)
death_1 = pd.Series(np.nan, index=heart_processed_df.index, dtype=float)

known_target = heart_processed_df['DEATH_EVENT'].notna()
death_0.loc[known_target] = (heart_processed_df.loc[known_target, 'DEATH_EVENT'] == 0).astype(int)
death_1.loc[known_target] = (heart_processed_df.loc[known_target, 'DEATH_EVENT'] == 1).astype(int)

heart_final_df = heart_processed_df.drop(columns=['DEATH_EVENT']).copy()
heart_final_df['DEATH_EVENT_0'] = death_0
heart_final_df['DEATH_EVENT_1'] = death_1

heart_final_df.to_csv('heart_failure_dataset_processed.csv', index=False)
print(heart_final_df.head())
print('Saved heart_failure_dataset_processed.csv')
```

## **Associations**

### Problem 2.5

```python
# Pearson correlations using the original, unprocessed data.
death_corr = (
    heart_raw_df.corr(method='pearson')['DEATH_EVENT']
    .drop('DEATH_EVENT')
    .sort_values(ascending=False)
)
print(death_corr)
```

#### (a)

The two variables most positively correlated with `DEATH_EVENT` are:

1. `serum_creatinine`: approximately **0.294**
2. `age`: approximately **0.254**


#### (b)

The two variables most negatively correlated with `DEATH_EVENT` are:

1. `time`: approximately **-0.527**
2. `ejection_fraction`: approximately **-0.269**


#### (c)

The second-most positive association is with `age`. A possible explanation is that older patients in this heart-failure cohort may generally have greater physiological vulnerability and more severe accompanying health problems, so death during follow-up occurs more often among older patients. This is an association in this dataset and does not by itself prove that age causes the outcome.


#### (d)

For `time`, the negative association is expected because `time` is the follow-up duration: follow-up ends when death occurs or at the last observation. Patients who die earlier therefore tend to have smaller `time` values, while patients who remain alive can accumulate longer follow-up times. This also means `time` is closely tied to how the outcome is recorded, so the correlation should not be interpreted as a simple causal effect.

For `ejection_fraction`, the variable measures the percentage of blood leaving the heart during each contraction. A lower ejection fraction indicates poorer pumping function. Therefore, lower ejection-fraction values being associated with a higher frequency of death produces a negative correlation with `DEATH_EVENT`. Correlation alone does not establish causation.
